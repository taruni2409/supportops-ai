```python
clean_df = df.copy()

print("Working copy created.")
```

```python
clean_df.columns.tolist()
```

```python
clean_df.columns = (
    clean_df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
```

```python
clean_df.columns.tolist()
```

```python
missing_count = clean_df.isnull().sum()

missing_percentage = (
    clean_df.isnull().sum() / len(clean_df) * 100
)

missing_summary = pd.DataFrame({
    "missing_count": missing_count,
    "missing_percentage": missing_percentage
})

missing_summary.sort_values(
    by="missing_percentage",
    ascending=False
)
```

```python
clean_df[
    ["ticket_subject", "ticket_description", "ticket_type"]
].isnull().sum()
```

```python
critical_missing = clean_df[
    clean_df["ticket_description"].isna()
    | clean_df["ticket_type"].isna()
]

print("Rows missing critical ML information:", len(critical_missing))
```

```python
clean_df = clean_df.dropna(
    subset=["ticket_description", "ticket_type"]
).copy()
```

```python
print("Shape after removing unusable NLP rows:", clean_df.shape)
```

```python
duplicate_descriptions = clean_df.duplicated(
    subset=["ticket_description"]
).sum()

print("Duplicate ticket descriptions:", duplicate_descriptions)
```

```python
print("Fully duplicated rows:", clean_df.duplicated().sum())
```

```python
description_label_counts = (
    clean_df
    .groupby("ticket_description")["ticket_type"]
    .nunique()
)

conflicting_descriptions = description_label_counts[
    description_label_counts > 1
]

print(
    "Descriptions associated with multiple ticket types:",
    len(conflicting_descriptions)
)
```

```python
if len(conflicting_descriptions) > 0:
    example_description = conflicting_descriptions.index[0]

    display(
        clean_df[
            clean_df["ticket_description"] == example_description
        ][
            ["ticket_description", "ticket_type"]
        ]
    )
```

```python
def basic_text_cleaning(text):
    text = str(text)
    
    # Remove leading/trailing spaces
    text = text.strip()
    
    # Replace multiple whitespace characters with one space
    text = " ".join(text.split())
    
    return text
```

```python
clean_df["ticket_description"] = (
    clean_df["ticket_description"]
    .apply(basic_text_cleaning)
)

clean_df["ticket_subject"] = (
    clean_df["ticket_subject"]
    .fillna("")
    .apply(basic_text_cleaning)
)
```

```python
empty_descriptions = (
    clean_df["ticket_description"].str.len() == 0
).sum()

print("Empty descriptions:", empty_descriptions)
```

```python
clean_df["ticket_text"] = (
    clean_df["ticket_subject"].fillna("")
    + " "
    + clean_df["ticket_description"].fillna("")
)
```

```python
clean_df["ticket_text"] = (
    clean_df["ticket_text"]
    .apply(basic_text_cleaning)
)
```

```python
clean_df[
    ["ticket_subject", "ticket_description", "ticket_text"]
].head()
```

```python
clean_df["text_character_count"] = (
    clean_df["ticket_text"].str.len()
)

clean_df["text_word_count"] = (
    clean_df["ticket_text"].str.split().str.len()
)
```

```python
clean_df[
    ["text_character_count", "text_word_count"]
].describe()
```

Protect against Personally Identifiable Information

```python
pii_columns = [
    "customer_name",
    "customer_email"
]

existing_pii_columns = [
    col
    for col in pii_columns
    if col in clean_df.columns
]

clean_df = clean_df.drop(
    columns=existing_pii_columns
)
```

```python
clean_df.columns.tolist()
```

Creating the ML-ready subset

```python
model_df = clean_df[
    [
        "ticket_id",
        "ticket_text",
        "ticket_type"
    ]
].copy()
```

```python
model_df.head()
```

```python
print("Rows:", model_df.shape[0])
print("Columns:", model_df.shape[1])

print("\nMissing values:")
print(model_df.isnull().sum())

print("\nDuplicate rows:")
print(model_df.duplicated().sum())

print("\nTarget classes:")
print(model_df["ticket_type"].value_counts())
```

```python
CLEAN_DATA_PATH = "../data/processed/customer_support_tickets_clean.csv"

clean_df.to_csv(
    CLEAN_DATA_PATH,
    index=False
)

print(
    f"Cleaned dataset saved to: {CLEAN_DATA_PATH}"
)
```

```python
MODEL_DATA_PATH = "../data/processed/ticket_classification_dataset.csv"

model_df.to_csv(
    MODEL_DATA_PATH,
    index=False
)

print(
    f"ML dataset saved to: {MODEL_DATA_PATH}"
)
```
